# Sign Language Symptom Detector

This notebook inspects the preprocessed dataset and explains the local training workflow. Each sample is a 30-frame sequence of MediaPipe Holistic landmarks. The reusable loader, model, trainer, and webcam demo live in `sld/`.

Run the training command from the project root as shown at the end. Training is intentionally not started automatically when this notebook is opened.


````python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / "sld").is_dir():
    raise RuntimeError("Open this notebook from the project root or its notebooks/ directory.")
sys.path.insert(0, str(PROJECT_ROOT))

from sld.config import ACTIONS, DEFAULT_DATASET_PATH
from sld.data import load_dataset
````

## Load and validate the data

The loader checks every expected frame, validates the 1,662-value feature vector, loads files in numeric order, and normalizes the legacy `soar throat` folder name to `sore throat`.


````python
X, y = load_dataset(DEFAULT_DATASET_PATH)
print(f"Sequences: {len(X)}")
print(f"Input shape: {X.shape}")
print(f"Label shape: {y.shape}")
print(f"Feature dtype: {X.dtype}")
print("
Sequences per class:")
for index, label in enumerate(ACTIONS):
    print(f"{label}: {(y == index).sum()}")
````

## Model input

The dataset contains 270 sequences. Each sequence has 30 frames, and each frame has 1,662 landmarks/features. The trainer uses a stratified split with a fixed seed so that all nine classes appear in both partitions.

## Train the model

From a terminal in the project root, run:

```bash
python -m sld.train
```

The command uses early stopping and saves the model under `models/`. See `python -m sld.train --help` for options.

## Run webcam inference

After training, run:

```bash
python -m sld.realtime
```

The demo requires a working camera and a trained model. Press **Q** or **Escape** to close it.
